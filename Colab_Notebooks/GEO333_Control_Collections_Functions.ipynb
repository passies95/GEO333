## 1. Comparison and logical operators
Comparisons return `True` or `False`: `==`, `!=`, `<`, `>`, `<=`, `>=`. Logical operators combine conditions: `and` requires both conditions true; `or` requires at least one; `not` reverses a Boolean.

````python
entrance_count = 2
is_public = True
print(entrance_count > 0)
print(entrance_count >= 2 and is_public)
print(entrance_count == 0 or not is_public)
````

**Expected output**
```text
True
True
False
```

## 2. Conditional statements
`if` checks the first condition. It check that a condition is true based on a statement. Often it integrates operators. `elif` checks another confition if earlier conditions were false, and `else` handles the remaining case. Indentation defines each block.

````python
entrance_count = 0

if entrance_count == 0:
    print("No entrance has been recorded.")
elif entrance_count == 1:
    print("One entrance has been recorded.")
else:
    print("Multiple entrances have been recorded.")
````

**Expected output:** `No entrance has been recorded.` This describes the current record; it does not prove the building physically has no entrance. The survey may be incomplete.

## 3. A simple record check
This checks that an ID is non-empty and floor count is positive. It is only a basic logical check, not field verification.

````python
building_id = "BLD-014"
floor_count = 3

if building_id != "" and floor_count > 0:
    print("Basic record check passed.")
else:
    print("Check the building ID and floor count.")
````

**Expected output:** `Basic record check passed.`

## 4. Repetition with a `for` loop
A `for` loop processes each item in a collection.

````python
building_names = ["Main Library", "Sports Centre", "Senate Building"]
for building in building_names:
    print("Building:", building)
````

**Expected output**
```text
Building: Main Library
Building: Sports Centre
Building: Senate Building
```

## 5. Repetition with `range()`
`range(n)` produces integers from zero up to, but not including, `n`. Thus `range(1, 4)` produces 1, 2 and 3.

````python
for number in range(1, 4):
    print("Record", number, "is ready for review.")
````

**Expected output**
```text
Record 1 is ready for review.
Record 2 is ready for review.
Record 3 is ready for review.
```

## 6. A `while` loop
A `while` loop repeats while its condition is true. Ensure something changes the condition so the loop eventually stops.

````python
reviewed = 0
while reviewed < 3:
    reviewed = reviewed + 1
    print("Reviewed record:", reviewed)
````

**Expected output**
```text
Reviewed record: 1
Reviewed record: 2
Reviewed record: 3
```

## 7. Lists, tuples and dictionaries
- **List:** ordered and changeable; useful for a sequence of elements such as names. It takes all the other basic data types, lists
- **Tuple:** ordered and immutable; useful for a fixed pair of illustrative coordinate values.
- **Dictionary:** key-value pairs; can be used for instance to record structured building record.

These are Python structures

````python
building_names = ["Main Library", "Sports Centre", "Senate Building"]
building_names.append("Faculty Building")
print(building_names)

sample_coordinate = (6.515, 3.389)
print(sample_coordinate)

building = {
    "building_id": "BLD-001",
    "building_name": "Main Library",
    "floor_count": 4,
    "is_public": True
}
print(building["building_name"])
print(building["floor_count"])
````

**Expected output**
```text
['Main Library', 'Sports Centre', 'Senate Building', 'Faculty Building']
(6.515, 3.389)
Main Library
4
```
The tuple is illustrative only. Real coordinates must be interpreted with a specified CRS and axis order.

## 8. A list of dictionaries: several records
A dictionary represents one simple record; a list can hold several records.

````python
buildings = [
    {"building_id": "BLD-001", "building_name": "Main Library", "floor_count": 4},
    {"building_id": "BLD-002", "building_name": "Sports Centre", "floor_count": 2},
    {"building_id": "BLD-003", "building_name": "Senate Building", "floor_count": 3}
]

for record in buildings:
    print(record["building_id"], "-", record["building_name"])
````

**Expected output**
```text
BLD-001 - Main Library
BLD-002 - Sports Centre
BLD-003 - Senate Building
```

## 9. Functions, parameters and return values
`def` defines a function. A **parameter** is a name in the definition that receives a value; an **argument** is the value supplied when calling it. `return` sends a result back to the calling code. `print()` displays; `return` passes a value for later use.

````python
def entrance_status(entrance_count):
    if entrance_count == 0:
        return "No entrance recorded"
    elif entrance_count == 1:
        return "One entrance recorded"
    else:
        return "Multiple entrances recorded"

print(entrance_status(0))
print(entrance_status(1))
print(entrance_status(3))
````

**Expected output**
```text
No entrance recorded
One entrance recorded
Multiple entrances recorded
```

## 10. Combine a loop, condition, list and function
Using our buidling examples, we may want for instance to idenfity building IDs of records with zero recorded water hydrants. This can the allow for instance a team to be sent to verify and mark the building for future improvements by adding a water hydrant. 

````python
campus_buildings = [
    {"building_id": "BLD-001", "name": "Main Library", "water_hydrant_count": 2},
    {"building_id": "BLD-002", "name": "Sports Centre", "water_hydrant_count": 0},
    {"building_id": "BLD-003", "name": "Senate Building", "water_hydrant_count": 1}
]

def find_records_without_water_hydrants(records):
    flagged_ids = []
    for record in records:
        if record["water_hydrant_count"] == 0:
            flagged_ids.append(record["building_id"])
    return flagged_ids

print(find_records_without_water_hydrants(campus_buildings))
````

**Expected output:** `['BLD-002']`

## 11. Guided practice — find incomplete records
Add a fourth record. Modify the function so it flags a record when its building ID is blank **or** its name is blank.

````python
group_records = [
    {"building_id": "BLD-011", "name": "Faculty of Science"},
    {"building_id": "BLD-012", "name": "Main Auditorium"},
    {"building_id": "", "name": "Unnamed building"}
]

def find_incomplete_records(records):
    incomplete = []
    for record in records:
        if record["building_id"] == "" or record["name"] == "":
            incomplete.append(record)
    return incomplete

print(find_incomplete_records(group_records))
````

**Expected output**
```text
[{'building_id': '', 'name': 'Unnamed building'}]
```
