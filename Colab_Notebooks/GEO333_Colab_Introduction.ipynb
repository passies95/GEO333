Choose **+ Code** or **+ Text** to add a cell. Click inside a cell to edit it. Run a code cell with its play icon or `Shift + Enter`. When later cells depend on earlier variables, run cells in order.

## 4. Comments
A comment is a note for people reading the code. It begins with `#`; Python ignores it when running the cell.

````python
# This is a comment: Python does not display it.
print("This message is displayed.")
# The next line describes a feature.
print("Feature type: Building")
````

**Expected output**
```text
This message is displayed.
Feature type: Building
```

## 5. A small campus/geography exercise
A building footprint (usually a polygon) and an entrance (usually a point) are different geographic features. One building can have multiple entrances.

````python
feature_name = "Senate Building"
feature_type = "Building"
visitor_note = "A campus landmark that visitors can search for."

print("Feature:", feature_name)
print("Type:", feature_type)
print("Visitor information:", visitor_note)
````

**Expected output**
```text
Feature: Main Library
Type: Building
Visitor information: A campus landmark that visitors can search for.
```
