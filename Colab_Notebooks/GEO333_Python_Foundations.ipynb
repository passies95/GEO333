## 1. Statements and expressions
An **expression** produces a value. A **statement** is an instruction executed by Python. `4 + 5` is an expression; `building_count = 5` is an assignment statement.

````python
4 + 5
````

**Expected output:** `9`. This cell will be displayed automatically in a notebook.

````python
building_count = 5
print(building_count)
````

**Expected output:** `5`.

## 2. Variables and naming
A variable is a name that refers to a value. They include names such as `building_name` and `entrance_count`. Names can contain letters, digits and underscores, but cannot begin with a digit or contain spaces.
A variable name should not begin with a number. 
Python is case-sensitive hence Station is different from station 
Do not use reserved words such as `if` or `for` as variable names. Reserved means they are used as part of core variables in python

````python
building_name = "Arts Block"
entrance_count = 1
is_public_building = True

print(building_name)
print(entrance_count)
print(is_public_building)
````

**Expected output**
```text
Arts Block
1
True
```

## 3. Four basic data types

| Type | Meaning | Example |
|---|---|---|
| `str` | Text | `"Main Library"` |
| `int` | Whole number | `3` |
| `float` | Decimal-capable number | `125.5` |
| `bool` | `True` or `False` | `True` |

Use `type()` to inspect a value. Text needs quotation marks; Boolean values use `True` or `False` without quotation marks.

````python
building_name = "Main Library"
entrance_count = 3
building_area_m2 = 125.5
is_accessible = True

print(type(building_name))
print(type(entrance_count))
print(type(building_area_m2))
print(type(is_accessible))
````

**Expected output**
```text
<class 'str'>
<class 'int'>
<class 'float'>
<class 'bool'>
```
An identifier such as `"BLD-001"` should be text, not a number: it identifies a feature rather than representing a quantity.

## 4. Operators
Arithmetic operators: `+`, `-`, `*`, `/`, `//`, `%`, `**`. Assignment uses `=`; compound assignment includes `+=`. Comparison operators (`==`, `!=`, `<`, `>`, `<=`, `>=`) will be used more fully in Week 3.

````python
building_count = 8
entrance_count = 14
total_features = building_count + entrance_count

print("Total features:", total_features)
print("Difference:", entrance_count - building_count)
print("Average entrances per building:", entrance_count / building_count)

reviewed_count = 0
reviewed_count += 1
print("Reviewed:", reviewed_count)
````

**Expected output**
```text
Total features: 22
Difference: 6
Average entrances per building: 1.75
Reviewed: 1
```

## 5. Type conversion
`input()` returns text (`str`). Convert before arithmetic: `int("4")` gives integer `4`; `float("125.75")` gives a decimal; `str(4)` gives text `"4"`. `int("four")` raises a `ValueError`.

````python
entrance_text = "4"
entrance_number = int(entrance_text)
area_text = "125.75"
area = float(area_text)

print(entrance_number + 2)
print(area)
print(str(entrance_number) + " entrances")
````

**Expected output**
```text
6
125.75
4 entrances
```

## 6. Input and formatted output
`input()` asks the user for information. An f-string (an `f` before the quotation mark) inserts values inside `{ }`. Run the cell and enter a building name and a whole number of entrances.

````python
building_name = input("Enter a building name: ")
entrances = int(input("Enter number of entrances: "))

print(f"{building_name} has {entrances} recorded entrances.")
print("One more would make:", entrances + 1)
````

**Example interaction**
```text
Enter a building name: Main Library
Enter number of entrances: 2
Main Library has 2 recorded entrances.
One more would make: 3
```
If the user types `two`, `int()` cannot convert it. Later, conditions will help us check input.

## 7. Debugging
- `SyntaxError`: invalid code structure, e.g. missing quotation mark or bracket.
- `NameError`: undefined or misspelled name.
- `TypeError`: an operation is unsuitable for the supplied types, e.g. adding text to an integer.
- `ValueError`: a value cannot be converted as requested, e.g. `int("four")`.

Debug systematically: read the last error line; inspect the indicated code; check spelling, quotes, brackets, indentation and types; fix one issue; run again.

````python
# Corrected example: convert text before adding
entrance_text = "3"
entrance_count = int(entrance_text)
print(entrance_count + 1)
````

**Expected output:** `4`.
